## atac

```python
new_mat = sc.read_mtx('/pool1/qiaolibo/R_Project/peak_gene/Marderstein_Nature_2024/processed/filter_matrix/atac.mtx')
new_mat = new_mat.T
```

```python
new_cell = pd.read_csv('../actinn/query/new/raw/cell.csv', index_col=0)
new_cell
```

```python
data = sc.read_h5ad('./processed/dat.h5ad')
```

```python
new_mat.obs_names = new_cell['cell_id'].tolist()
new_mat.var_names = data.var_names.tolist()
```

```python
df = pd.DataFrame([x.split('-') for x in new_mat.obs_names], columns=['dataset', 'barcode', 'sample'])
df['sample'] = df['dataset'] + '-' + df['sample']
df.index = new_mat.obs_names.tolist()
df
```

```python
new_mat.obs = df
```

```python
data_new = ad.concat([data, new_mat])
data_new
```

```python
data_new.write_h5ad('./processed/data_new.h5ad')
```
