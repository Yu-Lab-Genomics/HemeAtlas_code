```python
# Processing for Chen_Nat_Methods_2022 dataset
rna <- readRDS('z_others/3_rna_atac_cell/data/Chen_Nat_Methods_2022-rna.rds')
cell_id <- rna@Dimnames[[2]] # Extract cell IDs from the RNA object

meta <- read.csv('Chen_Nat_Methods_2022/raw/GSM5396332_CD4cells.csv')

# Reconstruct cell IDs using metadata components to match the RNA object format
meta$barcode <- substr(rownames(meta), 7, 22)
meta$cell <- paste0('Chen_Nat_Methods_2022-', meta$barcode, '-', meta$Sample)

# Set the constructed cell IDs as row names for convenient filtering and sorting
rownames(meta) <- meta$cell

# Filter and reorder metadata to match the exact order of cell IDs in the RNA object
meta_final <- meta[cell_id,]

# Select only the relevant annotation column (Clusters)
meta_final <- meta_final['Clusters']

# Rename cluster labels (Map cluster IDs to cell types)
table(meta_final[,"Clusters"])
meta_final <- meta_final %>%
  mutate(Clusters = recode(Clusters,
                           "C1" = "Act",
                           "C2" = "Treg",
                           "C3" = "Th17",
                           "C4" = "TCM",
                           "C5" = "Th2",
                           "C6" = "Th1",
                           "C7" = "Uncom"
                           ))

# Reference notes on cell counts and types:
# Th1 (1562)
# Th2 (939)
# Th17 (1,855)
# Treg (583)
# TCM, central memory (2,512)
# Act., recently activated cells (116) 
# Uncom., uncommitted memory cells (905)

# Verify the frequency table (corresponding to C1-C7 mappings)
table(meta_final[,"Clusters"])

# Save the final processed metadata
saveRDS(meta_final, "/path/to/hemelas_project/z_others/4_cell_ann/data/dataset.rds")
```
